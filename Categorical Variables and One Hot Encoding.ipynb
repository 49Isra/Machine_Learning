```python
from sklearn.preprocessing import OneHotEncoder 
import pandas as pd 

data =pd.DataFrame ({
    "Education" : ["master","degree","diploma","master","degree"] , 
    "Department" : ["IT","HR","Finance","IT","HR"] , 
    "City" : ["Delhi","Mumbai","Delhi","Bengalore","Mumbai"] 

 }) 

encoder = OneHotEncoder()
after = encoder.fit_transform(data[["Education","Department", "City"]]) 

print(after.toarray())
```

## Label Encoding

```python
from sklearn.preprocessing import LabelEncoder 
import pandas as pd 

data = pd.DataFrame({
    "Size" : ["Small","medium","Large"]
})

encoder = LabelEncoder()
size_encoded = encoder.fit_transform(data["Size"]) 

print(data)
```

## Target encoding 

```python
import pandas as pd 

data = pd.DataFrame ({
    "City" : ["Delhi","Mumbai","Delhi","Mumbai"], 
    "Price" : [500000,800000,550000,850000]
})

encoded = data.groupby("City")["Price"].mean() 
data["city_encoded"] = data["City"].map(encoded) 
print(data)

```
